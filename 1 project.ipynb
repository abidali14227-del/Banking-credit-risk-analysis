# Credit Risk Feature Engineering

#
Customer Financial Profile

#
Annual Income Band

```python
def income_band(income):
    if income < 300000:
        return "Low Income"
    elif income < 600000:
        return "Lower Middle Income"
    elif income < 1200000:
        return "Middle Income"
    elif income < 2500000:
        return "Upper Middle Income"
    else:
        return "High Income"

customers_clean["income_band"] = (
    customers_clean["annual_income"]
    .apply(income_band)
)
```

# Age Segmentation

```python
def age_group(age):
    if age < 25:
        return "18-24"
    elif age < 35:
        return "25-34"
    elif age < 45:
        return "35-44"
    elif age < 55:
        return "45-54"
    elif age < 65:
        return "55-64"
    else:
        return "65+"

customers_clean["age_group"] = (
    customers_clean["age"]
    .apply(age_group)
)
```

# DTI — Debt Burden

```python
def dti_band(dti):
    if dti < 20:
        return "Low DTI"
    elif dti < 40:
        return "Moderate DTI"
    elif dti < 60:
        return "High DTI"
    else:
        return "Very High DTI"

defaults_credit_risk_clean["dti_band"] = (
    defaults_credit_risk_clean["debt_to_income_pct"]
    .apply(dti_band)
)
```

# Credit Score Band

```python
def credit_score_band(score):
    if score < 580:
        return "Poor"
    elif score < 670:
        return "Fair"
    elif score < 740:
        return "Good"
    elif score < 800:
        return "Very Good"
    else:
        return "Exceptional"

defaults_credit_risk_clean["credit_score_band"] = (
    defaults_credit_risk_clean["credit_score"]
    .apply(credit_score_band)
)
```

# Payment Stress

# Missed Payment Band

```python
def missed_payment_band(x):
    if x == 0:
        return "No Missed Payment"
    elif x == 1:
        return "1 Missed Payment"
    elif x <= 3:
        return "2-3 Missed Payments"
    else:
        return "4+ Missed Payments"

defaults_credit_risk_clean["missed_payment_band"] = (
    defaults_credit_risk_clean["missed_payments_12m"]
    .apply(missed_payment_band)
)
```

# Delinquency Band

```python
def delinquency_band(days):
    if days == 0:
        return "No Delinquency"
    elif days <= 30:
        return "1-30 Days"
    elif days <= 60:
        return "31-60 Days"
    elif days <= 90:
        return "61-90 Days"
    else:
        return "90+ Days"

defaults_credit_risk_clean["delinquency_band"] = (
    defaults_credit_risk_clean["delinquency_days"]
    .apply(delinquency_band)
)
```

# Credit Card Utilization

```python
def utilization_band(x):
    if x < 30:
        return "Low Utilization"
    elif x < 50:
        return "Moderate Utilization"
    elif x < 75:
        return "High Utilization"
    else:
        return "Very High Utilization"

credit_cards_clean["utilization_band"] = (
    credit_cards_clean["utilization_pct"]
    .apply(utilization_band)
)
```

# Loan Exposure

```python
loans_clean["outstanding_ratio"] = (
    loans_clean["outstanding_amount"]
    / loans_clean["loan_amount"]
) * 100
```

# Loan Exposure Band

```python
def exposure_band(x):
    if x < 100000:
        return "Low Exposure"
    elif x < 500000:
        return "Medium Exposure"
    elif x < 1000000:
        return "High Exposure"
    else:
        return "Very High Exposure"

loans_clean["exposure_band"] = (
    loans_clean["outstanding_amount"]
    .apply(exposure_band)
)
```
